```python
row = spark.sql("""

SELECT
  secret('kv-northmart-gmkng', 'northmart-sql-user')     AS sql_user,
  secret('kv-northmart-gmkng', 'northmart-sql-password') AS sql_password
""").first()

sql_user = row["sql_user"]
sql_password = row["sql_password"]

print("User récupéré :", sql_user)
print("Password récupéré :", sql_password is not None)
print("Longueur password :", len(sql_password))
```
